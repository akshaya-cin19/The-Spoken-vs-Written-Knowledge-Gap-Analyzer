```python

import re
from difflib import SequenceMatcher

def tokenize(text):
    return set(re.findall(r"\b[a-zA-Z]+\b", text.lower()))

def analyze_answers(written, spoken, concepts):
    written_words = tokenize(written)
    spoken_words = tokenize(spoken)

    written_concepts = [
        c for c in concepts if c.lower() in written.lower()
    ]
    spoken_concepts = [
        c for c in concepts if c.lower() in spoken.lower()
    ]

    missing_speech = [
        c for c in written_concepts if c not in spoken_concepts
    ]

    common = written_words & spoken_words
    all_words = written_words | spoken_words

    overlap = len(common) / len(all_words) * 100 if all_words else 0
    similarity = SequenceMatcher(
        None, written.lower(), spoken.lower()
    ).ratio() * 100

    written_coverage = (
        len(written_concepts) / len(concepts) * 100 if concepts else 0
    )
    spoken_coverage = (
        len(spoken_concepts) / len(concepts) * 100 if concepts else 0
    )

    print("=" * 55)
    print("SPOKEN VS WRITTEN KNOWLEDGE GAP ANALYZER")
    print("=" * 55)
    print("Question: Explain photosynthesis.")
    print("\nWritten concepts covered:", ", ".join(written_concepts))
    print("Spoken concepts covered: ", ", ".join(spoken_concepts))
    print("Missing from spoken answer:",
          ", ".join(missing_speech) if missing_speech else "None")
    print(f"Written concept coverage: {written_coverage:.1f}%")
    print(f"Spoken concept coverage:  {spoken_coverage:.1f}%")
    print(f"Unique-word overlap:      {overlap:.1f}%")
    print(f"Text similarity:          {similarity:.1f}%")

    print("\nSuggestion:")
    if missing_speech:
        print("Revise these concepts before speaking:",
              ", ".join(missing_speech))
    else:
        print("All written concepts are covered in the speech.")
    print("=" * 55)


# Sample input
written_answer = (
    "Photosynthesis is the process by which green plants use sunlight, "
    "carbon dioxide, and water to produce glucose and oxygen. "
    "Chlorophyll absorbs light energy."
)

spoken_transcript = (
    "Photosynthesis is how green plants use sunlight and water "
    "to make glucose. Chlorophyll absorbs light energy."
)

key_concepts = [
    "sunlight", "carbon dioxide", "water",
    "glucose", "oxygen", "chlorophyll"
]

analyze_answers(
    written_answer,
    spoken_transcript,
    key_concepts
)
```
